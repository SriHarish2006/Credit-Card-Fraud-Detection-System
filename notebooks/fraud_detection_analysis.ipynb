# Credit Card Fraud Detection — Exploratory Data Analysis

This notebook explores the credit-card transactions dataset, checks data quality,
analyzes class imbalance, and visualizes fraud patterns before model training.

Run this notebook from the project root (so relative paths like `data/creditcard.csv` resolve).

````
import sys, os
sys.path.append(os.path.abspath(".."))  # allow `import src...` when run from notebooks/

import pandas as pd
from src import eda
from src.data_preprocessing import load_dataset, validate_dataset, profile_dataset, clean_dataset

DATA_PATH = os.path.join("..", "data", "creditcard.csv")
````

## 1. Load and validate the dataset

````
df_raw = load_dataset(DATA_PATH)
validate_dataset(df_raw)
report = profile_dataset(df_raw)

print(f"Rows: {report.n_rows}, Columns: {report.n_cols}")
print(f"Duplicate rows: {report.duplicate_rows}")
print(f"Total missing values: {sum(report.missing_values.values())}")
print(f"Genuine: {report.genuine_count} ({report.genuine_percentage}%)")
print(f"Fraud: {report.fraud_count} ({report.fraud_percentage}%)")
````

## 2. Clean the dataset (handle duplicates / missing values)

````
df = clean_dataset(df_raw)
df.shape
````

## 3. Fraud statistics

Fraud detection datasets like this one are **highly imbalanced** — fraud typically
makes up well under 1% of all transactions. This has major implications for how we
train and evaluate models later (see class-imbalance handling in `train_model.py`).

````
eda.fraud_statistics(df)
````
